### Project Requirements 

""" 
Take a user question

Send it to an LLM via LangChain

Only Return a structured answer for only tech questions 

Also don't answer if the question of the user is not relavent to tech 

"""

```python
from langchain_core.prompts import ChatPromptTemplate

prompt_template = ChatPromptTemplate.from_messages([
    ("system" , "you are a question reply assistant also assume u as a profossor in ai tech also if the user asks anything not related to ai and tech just say '''i am only allowed to give answer based on the tech field only sorry for that so now can we take about any new tech or AI''' and you are supposed to give only the clear cut main core output for the user queries also answer in the professional teacher who explain in simple words in small lines ") ,
    ("user" , "Explain the concepts of {text} in simple words ")
])

user_inputs = input("Enter a topic that u need to learn")

ready_prompt = prompt_template.invoke({"text" : user_inputs})

prompt = ready_prompt.messages[1].content

prompt
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm_openai = ChatGoogleGenerativeAI(model="gemini-2.5-flash" , temperature = 0.9 )

llm_openai.invoke(ready_prompt).content
```
